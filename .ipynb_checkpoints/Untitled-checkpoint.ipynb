```python
import numpy as np
```

```python
import pandas as pd
```

```python
df = pd.read_csv('foodmandu_data.csv')
```

```python
df.info()
```

```python
df['Price'] = df['Price'].str.replace(',', '')
```

```python
df['Price'] = df['Price'].astype(float)
```

```python
df['Price'] = df['Price'].fillna(0)
```

```python
df['Price'] = df['Price'].astype(int)
```

```python
rn = df['Restaurant Name'].value_counts()
```

```python
keys = dict(rn).keys()
values = dict(rn).values()
```

```python
# import matplotlib as pl
import matplotlib.pyplot as pl
```

```python
# steps:
#     1.figure size
#     2.actual chart type() bar graph, pie chart
#     3.Xlabel(NAME)
#     4.Ylabel(Name)
#     5.Title of a chart
#     6.Display the chart
```

```python
pl.figure(figsize=(15,5))
pl.title('Restaurant and its count')
pl.bar(keys,values)
pl.xlabel("Name of restaurats")
pl.xlabel("Counts of restaurants")
pl.xticks(rotation=90)
```
