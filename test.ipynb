```python
df = pd.DataFrame({
    "Student": ["Ram", "Sita", "Hari", "Gita", "Shyam"],
    "Grade": ["B", "A", "C", "A+", "B+"]
})
```

```python
grade_mapping = {
    "C": 1,
    "B": 2,
    "B+": 3,
    "A": 4,
    "A+": 5
}
```

```python
df["Grade"] = df["Grade"].map(grade_mapping)
df
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[["C", "B", "B+", "A", "A+"]]
)

df[["Grade"]] = encoder.fit_transform(df[["Grade"]])
```
