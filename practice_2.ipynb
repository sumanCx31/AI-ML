```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('nepal_students_dataset.csv')
```

```python
df
```

```python
df['study_hours_per_week']
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(df.study_hours_per_week, df['math_marks'], color='blue', s=2)

plt.title('Scatter Plot of Study hoours vs Math marks')
plt.xlabel('Study hours per week')
plt.ylabel('Marks in math')
plt.grid(True)
```
