```python
df = pd.read_csv('dataset/foodmandu_data.csv')
```

```python
df['Cuisine Type'].value_counts().head(6)
```

```python
df.head()
```

```python
top_cuisines = df['Cuisine Type'].value_counts().head(10)
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    top_cuisines.values,
    labels=top_cuisines.index,
    startangle=150,
    colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6']
)

plt.title('Top Cuisine Types Distribution in Foodmandu Data')
plt.axis('equal') 
plt.tight_layout()
plt.show()
```

```python
df['Price_clean'] = df['Price'].str.replace(',', '').astype(float)
```

```python
top_restaurants = df.groupby('Restaurant Name')['Price_clean'].mean().sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10, 5))
top_restaurants.plot(kind='line', marker='o', color='b', linestyle='-', linewidth=2)

plt.title('Average Food Price for Top 10 Highest-Priced Restaurants')
plt.ylabel('Average Price (NPR)')
plt.xlabel('Restaurant Name')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
```
