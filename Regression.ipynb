```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame({'Qty':[1,2,4,5,6],'Price':[80,160,320,400,480]})
```

```python
m=50
c=2
df['T1']=m*df['Qty']+c
df['T2']=(m+5)*df['Qty']+c+1
df['T3']=df['Qty']+0
df['T4']=(m+9)*df['Qty']+c+3
df['T4']=(m+12)*df['Qty']+c+4
df['T5']=(m+11)*df['Qty']+c+7
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(df['Qty'], df['Price'], color='blue',s=20)
plt.plot(df['Qty'],df['Price'],color='blue')
plt.plot(df.T2,df['Price'],color='red')
plt.title('Scatter Plot of Quantity vs  Price')
plt.xlabel('Quantity')
plt.ylabel('Price')
plt.grid(True)
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(df.T1, df['Price'], color='orange',s=20)
# plt.plot(df.Qty,df['Price'],color='blue')
plt.plot(df.T1,df['Price'],color='orange')
plt.plot(df.T2,df['Price'],color='red')
plt.plot(df.T3,df['Price'],color='blue')
plt.plot(df.T4,df['Price'],color='green')
plt.plot(df.T5,df['Price'],color='yellow')
plt.title('Scatter Plot of Quantity vs  Price')
plt.xlabel('Quantity')
plt.ylabel('Price')
plt.grid(True)
```
